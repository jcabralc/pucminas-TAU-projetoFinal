```python
import os
import sys
import pathlib

from dotenv import load_dotenv

RAIZ = pathlib.Path.cwd()
sys.path.insert(0, str(RAIZ))
load_dotenv(RAIZ / ".env")

from src.aura_client import AuraClient

base_url = os.environ["AURORA_BASE_URL"]
username = os.environ["AURORA_USERNAME"]
password = os.environ["AURORA_PASSWORD"]

client = AuraClient(base_url, username, password)
print(f"client criado")
```

```python
## Checagem rápida (cota esgotada)

import datetime

from src.aura_client import AuraQuotaExhausted, AuraServerError

agora = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

try:
    client.health()
    resposta = client.chat("Oi")
    print(f"[{agora}] OK, cota liberada! resposta: {resposta.get('message')}")
except AuraQuotaExhausted as erro:
    print(f"[{agora}] ainda esgotada: {erro}")
except AuraServerError as erro:
    print(f"[{agora}] erro interno/infra, tente de novo em instantes: {erro}")
```
